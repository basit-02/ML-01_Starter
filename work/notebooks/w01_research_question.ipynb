## 3. Quick look at the data (2-3 real numbers)

*To validate this lane, I looked at the trailing-90-day metrics in the starter dataset. I specifically accounted for the data gotcha that avg_position = 0 means "no data," not rank zero, to ensure my baseline metrics are honest.*

```python
import pandas as pd

# Load the starter dataset directly using the raw GitHub URL from repo
url = "https://raw.githubusercontent.com/basit-02/ML-01_Starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# 1. Total rows (one row per pseudonymized content item)
total_pages = len(df)

# 2. Percentage of pages actively showing a downward trend
declining_pct = (df["trend_direction"] == "down").mean() * 100

# 3. Valid pages missing position data (avg_position == 0 means "no data")
missing_position = len(df[df["avg_position"] == 0])

print(f"Total content items in starter dataset: {total_pages:,}")
print(f"Pages actively trending downward: {declining_pct:.1f}%")
print(f"Pages with missing position data (avg_position=0): {missing_position:,}")
```
