## 1. My lane as an ML task (type)

*This is fundamentally a Ranking/Scoring task, powered by binary classification probabilities. The human action is an editor reviewing a queue of pages to update. If we only use strict classification ("Is it declining? Yes/No"), we are left with thousands of "Yes" pages and no way to prioritize them. By outputting a probability score, we can rank the pages and tell the editor which one to fix first.*

```python
# No complex code needed here, just a print statement confirming the lane choice
print("Task Type: Ranking via Classification Probabilities")
print("Goal: Prioritize decaying pages for the editorial review queue.")
```

## 2. Target or proxy

*The target is the observed decline in a page's performance. The proxy we will use to predict this is a binary label indicating whether the observed trend_direction is 'down'. This is an observed historical outcome, not a hand-written rule like health_score.*

```python
import pandas as pd

# Load the starter dataset directly using the raw GitHub URL from your repo
url = "https://raw.githubusercontent.com/basit-02/ML-01_Starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Now construct the target
df["is_declining_label"] = (df["trend_direction"].str.lower() == "down").astype(int)

print(f"Target column created: 'is_declining_label'")
print(f"Base rate of declining pages: {df['is_declining_label'].mean():.2%}")
```

## 3. Success metric

*The success metric is Precision@K (e.g., Precision@50). The cost of a false positive is wasted editorial time, a human reviewing a page that doesn't actually need a refresh. Since editors work through a queue from top to bottom, we care deeply about the density of true positives at the very top of our recommendations. If we hand an editor 50 pages, what fraction are actually decaying?*

```python
# Define how you will measure success
print("Primary Metric: Precision@K")
print("Business translation: Out of the top K pages flagged by the model, how many actually require an editorial update?")
```

## 4. The unit of analysis, as a real dataframe

*The unit of analysis is a single web page evaluated over a specific time window. One row equals one unique page's performance snapshot.*

```python
# Show the unit of analysis
unit_df = df[['content_id', 'client_id', 'days_since_last_update', 'impressions_90d', 'is_declining_label']]
print(f"Dataset shape: {unit_df.shape[0]} pages")
display(unit_df.head(3))
```

## 5. Why ML beats a fixed rule here

*A fixed rule uses rigid, hand-written thresholds (e.g., days_since_last_update > 180 AND impressions_90d > 500). These hard cutoffs create a binary cliff. ML beats this because it evaluates multiple continuous signals simultaneously to find non-linear patterns. A high-value page updated 160 days ago might be far more urgent to refresh than a low-value page updated 190 days ago. ML can weight these trade-offs smoothly, providing a nuanced, directional score rather than a blunt filter.*

```python
# Show how a hard rule leaves out potentially valuable pages
hard_rule_misses = df[(df['days_since_last_update'] < 180) & (df['is_declining_label'] == 1)]
print(f"Pages declining but missed by a 180-day hard rule: {len(hard_rule_misses)}")
```
